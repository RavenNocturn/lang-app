```python
import kagglehub

# Download latest version
path = kagglehub.dataset_download("nezahatkk/10-000-english-words-cerf-labelled")

print("Path to dataset files:", path)
```

```python
import pandas as pd
```

```python
words = pd.read_csv(f"{path}/ENGLISH_CERF_WORDS.csv")
```

```python
words.head()
```

```python
words.info()
```

```python
level_counts = words.groupby("CEFR").size().reset_index(name="count")

level_counts["percentage"] = (
    level_counts["count"] / level_counts["count"].sum() * 100
).round(2)


print(level_counts)
```

```python
words.to_csv("words_cefr.csv", index=False)
```

```python
cefr_rank = {
    "A1": 1,
    "A2": 2,
    "B1": 3,
    "B2": 4,
    "C1": 5,
    "C2": 6
}

words["level_num"] = words["CEFR"].map(cefr_rank)
words.to_csv("words_level.csv", index=False)
```
